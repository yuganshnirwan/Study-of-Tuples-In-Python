#Study Of Tuples in Python

```python
# Create Tuple and Access Tuple Items
thistuple=("flower","leaf","stem")
print(thistuple[1])
```

```python
tuple1=('Python','c++','c','sql','r')
print(tuple1[0::3])
```

```python
# Replacing
thistuple=("flower","leaf","stem")
# Tuples are immutable, so direct item assignment is not allowed.
# To "replace" an item, convert to a list, modify, then convert back to a tuple.
list_from_tuple = list(thistuple)
list_from_tuple[1] = "roots"
thistuple = tuple(list_from_tuple)
print(thistuple)
```

```python
# Append Items
thistuple=("flower","leaf","stem")
# Tuples are immutable, so direct appending is not allowed.
# To "append" an item, convert to a list, modify, then convert back to a tuple.
list_from_tuple = list(thistuple)
list_from_tuple.append("roots")
thistuple = tuple(list_from_tuple)
print(thistuple)
```

```python
# Adding
t1=(4,7,2)
t2=(9,10,7)
t3= t1+t2
print("t1:",t1)
print("t2",t2)
print("t1+t2:",t3)
```

```python
# Result Storing
'''A students exam result is stored in a tuple containing three fixed fields:
Subject name,marks obtained and grade
write a python program to :
1. store the exam result using a tuple
2. unpack the tuple elements into seperate variables.
3. Display the suject, marks, grade.
4. Check whether the student has scored 75 or more marks and print "Distinction" if the condition is satisfied.'''

student=("Maths",89,"A")
sub=student[0]
marks=student[1]
grade=student[2]
print("Subject Name:",sub)
print("Marks:",marks)
print("Grade:",grade)

if marks>75:
  print("Distinction")
```
